```python
import wntr
import pandas as pd
from tqdm import tqdm  # progress bar

inp_path = "Net3.inp"

def simulate_leak(node_name, emitter_coeff=0.001):
    """Simulate a leak at a given node with a given emitter coefficient."""
    wn = wntr.network.WaterNetworkModel(inp_path)
    junction = wn.get_node(node_name)
    junction.emitter_coefficient = emitter_coeff
    
    sim = wntr.sim.EpanetSimulator(wn)
    results = sim.run_sim()
    
    pressure = results.node["pressure"].reset_index().rename(columns={"index": "Time"})
    pressure["Leak_Node"] = node_name
    pressure["Leak_Label"] = 1
    pressure["Leak_Size"] = emitter_coeff
    return pressure

# ---- 1. Baseline (no leak) ----
wn = wntr.network.WaterNetworkModel(inp_path)
sim = wntr.sim.EpanetSimulator(wn)
results = sim.run_sim()
baseline = results.node["pressure"].reset_index().rename(columns={"index": "Time"})
baseline["Leak_Node"] = "None"
baseline["Leak_Label"] = 0
baseline["Leak_Size"] = 0.0

# ---- 2. Loop over ALL junctions × leak sizes ----
leak_nodes = wn.junction_name_list
leak_sizes = [0.001, 0.005, 0.01]

all_data = [baseline]

for node in tqdm(leak_nodes, desc="Simulating nodes"):
    for size in leak_sizes:
        try:
            leak_df = simulate_leak(node, emitter_coeff=size)
            all_data.append(leak_df)
        except Exception as e:
            print(f"⚠️ Skipped node {node} (error: {e})")

# ---- 3. Combine everything ----
dataset = pd.concat(all_data, ignore_index=True)

# Replace NaN in Leak_Node with "None"
dataset["Leak_Node"] = dataset["Leak_Node"].fillna("None")

# ---- 4. Save final dataset ----
dataset.to_csv("net3_dataset_full.csv", index=False)

print("✅ Final dataset saved as net3_dataset_full.csv")
print("Shape:", dataset.shape)
print("Leak nodes included:", dataset['Leak_Node'].nunique(), "unique nodes")
print("Leak sizes included:", dataset['Leak_Size'].unique())
```

```python
import pandas as pd
import os

def log_results(model_name, task, report, summary_file="results_summary.csv"):
    """
    Save model results into a consolidated CSV.
    
    model_name: str (e.g., 'RandomForest')
    task: str ('Detection' or 'Localization')
    report: dict from classification_report(output_dict=True)
    """
    # Extract main metrics
    if task == "Detection":
        accuracy = report["accuracy"]
        precision = report["1"]["precision"]  # leak class
        recall = report["1"]["recall"]
        f1 = report["1"]["f1-score"]
        row = {
            "Model": model_name,
            "Task": task,
            "Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1": f1,
            "Macro_F1": "-",
            "Weighted_F1": "-"
        }
    else:  # Localization (multiclass)
        accuracy = report["accuracy"]
        macro_f1 = report["macro avg"]["f1-score"]
        weighted_f1 = report["weighted avg"]["f1-score"]
        row = {
            "Model": model_name,
            "Task": task,
            "Accuracy": accuracy,
            "Precision": "-",
            "Recall": "-",
            "F1": "-",
            "Macro_F1": macro_f1,
            "Weighted_F1": weighted_f1
        }
    
    # Append to summary file
    df_row = pd.DataFrame([row])
    if os.path.exists(summary_file):
        df_row.to_csv(summary_file, mode="a", header=False, index=False)
    else:
        df_row.to_csv(summary_file, mode="w", header=True, index=False)
    
    print(f"✅ Logged {model_name} ({task}) results to {summary_file}")
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score

def top_k_accuracy(y_true, y_prob, k=3):
    """
    Compute Top-k accuracy.
    y_true: array of true class labels
    y_prob: 2D array of class probabilities (n_samples × n_classes)
    """
    top_k_preds = np.argsort(y_prob, axis=1)[:, -k:]  # indices of top-k predictions
    hits = [y_true[i] in top_k_preds[i] for i in range(len(y_true))]
    return np.mean(hits)

def evaluate_topk(model, X_test, y_test, model_name, summary_file="results_topk.csv"):
    """
    Evaluate Top-1, Top-3, Top-5 accuracy for a trained model.
    Saves results into results_topk.csv
    """
    # Get probabilities
    if hasattr(model, "predict_proba"):  # scikit-learn models
        y_prob = model.predict_proba(X_test)
    else:  # Keras/TensorFlow models
        y_prob = model.predict(X_test)

    top1 = accuracy_score(y_test, np.argmax(y_prob, axis=1))
    top3 = top_k_accuracy(y_test, y_prob, k=3)
    top5 = top_k_accuracy(y_test, y_prob, k=5)

    row = pd.DataFrame([{
        "Model": model_name,
        "Top-1": top1,
        "Top-3": top3,
        "Top-5": top5
    }])

    if not os.path.exists(summary_file):
        row.to_csv(summary_file, index=False)
    else:
        row.to_csv(summary_file, mode="a", header=False, index=False)

    print(f"✅ {model_name} Top-k evaluation complete. Results saved to {summary_file}")
    return row
```

```python
import pandas as pd
from docx import Document
from docx.shared import Pt
from docx.oxml import OxmlElement
from docx.oxml.ns import qn

# --- Load results ---
summary = pd.read_csv("results_summary.csv")
topk = pd.read_csv("results_topk.csv")

# --- Merge into Combined view ---
combined = pd.merge(summary, topk, on="Model", how="outer")

# Ensure numeric where possible
for col in ["Accuracy","F1","Macro_F1","Weighted_F1","Top-1","Top-3","Top-5"]:
    if col in combined.columns:
        combined[col] = pd.to_numeric(combined[col], errors="coerce")

# Sort by Top-1
combined_sorted = combined.sort_values(by="Top-1", ascending=False)

# --- Find best per column ---
best_scores = {}
for col in ["Accuracy","F1","Macro_F1","Weighted_F1","Top-1","Top-3","Top-5"]:
    if col in combined_sorted.columns:
        best_scores[col] = combined_sorted[col].max()

# --- Create Word document ---
doc = Document()
doc.add_heading("Results Summary Tables", level=1)

# Helper to bold best values
def add_table_from_df(df, title, style="Light Grid Accent 3", highlight=True):
    doc.add_heading(title, level=2)
    table = doc.add_table(rows=1, cols=len(df.columns))
    table.style = style
    
    # Header
    hdr_cells = table.rows[0].cells
    for i, col in enumerate(df.columns):
        hdr_cells[i].text = col
    
    # Rows
    for _, row in df.iterrows():
        row_cells = table.add_row().cells
        for i, col in enumerate(df.columns):
            val = row[col]
            row_cells[i].text = str(val)
            # Highlight best values
            if highlight and col in best_scores and pd.notna(val) and val == best_scores[col]:
                run = row_cells[i].paragraphs[0].runs[0]
                run.font.bold = True
                # Optional shading (light yellow)
                tcPr = row_cells[i]._tc.get_or_add_tcPr()
                shd = OxmlElement('w:shd')
                shd.set(qn('w:fill'), "FFFF99")  # light yellow
                tcPr.append(shd)

# --- Table 1: Summary ---
add_table_from_df(summary, "Table 1. Detection and Localization Metrics", style="Light List Accent 1", highlight=False)

# --- Table 2: Top-k ---
add_table_from_df(topk, "Table 2. Top-k Localization Accuracy", style="Light List Accent 2", highlight=False)

# --- Table 3: Combined (sorted, highlighted best) ---
add_table_from_df(combined_sorted, "Table 3. Combined Metrics and Top-k Accuracy (Best highlighted)", style="Light Grid Accent 3", highlight=True)

# Save
doc.save("Results_Tables.docx")
print("✅ Word tables saved as Results_Tables.docx (with highlighted best scores)")
```

```python
import pandas as pd
import numpy as np
import joblib
import random
from sklearn.metrics import accuracy_score
from tensorflow.keras.models import load_model
import matplotlib.pyplot as plt

# --- Load dataset ---
df = pd.read_csv("net3_dataset_full.csv")

# ✅ Only use leak cases (localization task)
df = df[df["Leak_Label"] == 1].copy()
df["Leak_Node"] = df["Leak_Node"].fillna("Baseline").astype(str)

# Load encoder
encoder = joblib.load("leak_node_encoder.pkl")

# Feature matrix & labels
all_features = [c for c in df.columns if c not in ["Time","Leak_Node","Leak_Label","Leak_Size"]]
X_full = df[all_features].values
y_full = encoder.transform(df["Leak_Node"])

# --- Define models (already trained full models) ---
models = {
    "RandomForest": joblib.load("rf_localization_model.pkl"),
    "XGBoost": joblib.load("xgb_localization_model.pkl"),
    "LightGBM": joblib.load("lgb_localization_model.pkl"),
    "MLP": load_model("mlp_localization_model.h5"),
    "LSTM": load_model("lstm_localization_model.h5"),
    "CNN": load_model("cnn_localization_model.h5"),
}

# --- Helper for sequences (LSTM & CNN) ---
def create_sequences(X, y, timesteps=10):
    Xs, ys = [], []
    for i in range(len(X) - timesteps):
        Xs.append(X[i:(i+timesteps)])
        ys.append(y[i+timesteps])
    return np.array(Xs), np.array(ys)

# --- Sensor reduction experiment ---
results = []
sensor_drop_rates = [0.1, 0.3, 0.5]  # 10%, 30%, 50% sensors removed
n_features = len(all_features)

for drop_rate in sensor_drop_rates:
    n_keep = int(n_features * (1 - drop_rate))
    kept_features = random.sample(all_features, n_keep)
    
    X_reduced = df[kept_features].values
    y_reduced = encoder.transform(df["Leak_Node"])

    for model_name, model in models.items():
        if model_name in ["LSTM","CNN"]:  # sequence models
            if len(X_reduced) < 11:
                continue
            X_seq, y_seq = create_sequences(X_reduced, y_reduced, timesteps=10)
            y_prob = model.predict(X_seq)
            y_pred = np.argmax(y_prob, axis=1)
            acc = accuracy_score(y_seq, y_pred)
        else:
            if hasattr(model, "predict_proba"):
                y_pred = model.predict(X_reduced)
            else:
                y_pred = np.argmax(model.predict(X_reduced), axis=1)
            acc = accuracy_score(y_reduced, y_pred)
        
        results.append({
            "Model": model_name,
            "Sensor Coverage": f"{int((1-drop_rate)*100)}%",
            "Accuracy": acc
        })

# --- Save reference ---
df_results = pd.DataFrame(results)
df_results.to_csv("SensorReduction_Analysis.csv", index=False)
df_results.to_excel("SensorReduction_Analysis.xlsx", index=False)

# --- Plot ---
plt.figure(figsize=(10,6))
for model_name in df_results["Model"].unique():
    subset = df_results[df_results["Model"] == model_name]
    plt.plot(subset["Sensor Coverage"], subset["Accuracy"], marker="o", label=model_name)

plt.xlabel("Sensor Coverage")
plt.ylabel("Localization Accuracy")
plt.title("Localization Accuracy vs. Sensor Coverage")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("SensorReduction_Analysis.png", dpi=300)
plt.close()

print("✅ Figure saved as SensorReduction_Analysis.png")
print("✅ Reference data saved as SensorReduction_Analysis.csv and SensorReduction_Analysis.xlsx")
```

```python
# gnn_localization.py
import os
import numpy as np
import pandas as pd
from tqdm import tqdm
import joblib
import matplotlib.pyplot as plt

# PyTorch & PyG imports
import torch
import torch.nn.functional as F
from torch import nn, optim
from torch_geometric.data import Data, DataLoader
from torch_geometric.nn import GCNConv, SAGEConv, GATConv, global_mean_pool

# ---------- Config ----------
DATA_CSV = "net3_dataset_full.csv"
INP_PATH = "Net3.inp"   # used to extract topology via wntr if available
ENCODER_PKL = "leak_node_encoder.pkl"
OUT_DIR = "gnn_results"
BATCH_SIZE = 32
NUM_EPOCHS = 30         # adjust or increase for final experiments
LR = 1e-3
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
TIMESTEPS = 1           # using single snapshot; if using windows, adjust pipeline

os.makedirs(OUT_DIR, exist_ok=True)

# ---------- 1) Build/Load topology (edge_index) ----------
def build_edge_index_from_inp(inp_path):
    try:
        import wntr
    except Exception as e:
        print("wntr not available, cannot parse .inp automatically.", e)
        return None
    wn = wntr.network.WaterNetworkModel(inp_path)
    # build undirected edges from pipes (link endpoint names)
    edges = []
    for pipe_name in wn.pipe_name_list:
        pipe = wn.get_link(pipe_name)
        n1 = pipe.start_node_name
        n2 = pipe.end_node_name
        # convert names to indices as per wn.junction_name_list order
        # we will create mapping later to keep consistent
        edges.append((str(n1), str(n2)))
    return edges, wn.junction_name_list  # return raw edges and node list

# Try to extract topology
edge_pairs = None
node_order = None
try:
    res = build_edge_index_from_inp(INP_PATH)
    if res is not None:
        edge_pairs, node_order = res
        print(f"Loaded topology from {INP_PATH} ({len(edge_pairs)} edges, {len(node_order)} junctions).")
except Exception as e:
    print("Topology extraction failed:", e)

# ---------- 2) Load dataset & encoder ----------
df = pd.read_csv(DATA_CSV)
le_encoder = joblib.load(ENCODER_PKL)
num_classes = len(le_encoder.classes_)
print("Number of classes (nodes):", num_classes)

# Keep only leak samples for localization (we used that previously)
df_loc = df[df["Leak_Label"] == 1].copy()
df_loc["Leak_Node"] = df_loc["Leak_Node"].fillna("Baseline").astype(str)

# Identify node feature columns (pressure columns)
feat_cols = [c for c in df.columns if c not in ["Time","Leak_Node","Leak_Label","Leak_Size"]]
num_nodes = len(feat_cols)
print("Detected node feature columns:", num_nodes)

# Build mapping from node name -> column index:
# We assume your encoder fitted on Leak_Node used the same node names/strings as the column names.
# We will attempt to map encoder.classes_ order to dataset columns: if exact match, use that.
columns_set = set(feat_cols)
mapping_node_to_col = {}
for node_name in le_encoder.classes_:
    if node_name in columns_set:
        mapping_node_to_col[node_name] = feat_cols.index(node_name)

# If we do not have mapping (names differ), fall back to numeric string mapping assuming columns are node IDs
if len(mapping_node_to_col) == 0:
    # Try converting column names and encoder names to strings and match
    for node_name in le_encoder.classes_:
        for idx, col in enumerate(feat_cols):
            if str(col) == str(node_name):
                mapping_node_to_col[node_name] = idx

# If still incomplete, create default mapping using sorted order (last resort)
if len(mapping_node_to_col) != num_nodes:
    print("Warning: mapping between encoder classes and feature columns is incomplete."
          " Using dataset column order as node order. This assumes the encoding matches column order.")
    node_name_order = feat_cols[:]  # string names
else:
    # reorder node list to match columns
    node_name_order = feat_cols[:]  # use columns order

# ---------- 3) Build edge_index (PyG) using node_name_order ----------
# If we have extracted edges with WNTR, map edge node names to indices in node_name_order
if edge_pairs is not None and node_order is not None:
    name_to_idx = {str(n): i for i, n in enumerate(node_name_order)}
    edge_index_list = []
    for a, b in edge_pairs:
        if a in name_to_idx and b in name_to_idx:
            i = name_to_idx[a]
            j = name_to_idx[b]
            edge_index_list.append([i, j])
            edge_index_list.append([j, i])
    edge_index = torch.tensor(edge_index_list, dtype=torch.long).t().contiguous()
    print("Edge index built from INP:", edge_index.shape)
else:
    # fallback: fully connected ring or k-NN? We'll build simple chain edges using column neighbor order
    edge_index_list = []
    for i in range(num_nodes - 1):
        edge_index_list.append([i, i+1])
        edge_index_list.append([i+1, i])
    edge_index = torch.tensor(edge_index_list, dtype=torch.long).t().contiguous()
    print("Fallback edge_index built:", edge_index.shape)

# ---------- 4) Create PyG Data objects (one graph per sample) ----------
data_list = []
labels = []
for idx, row in tqdm(df_loc.iterrows(), total=len(df_loc), desc="Building graphs"):
    # node features: pressures in the order feat_cols
    x = torch.tensor(row[feat_cols].values.astype(np.float32)).unsqueeze(1)  # shape (num_nodes, 1)
    # make graph data
    y_node_name = row["Leak_Node"]
    y_label = int(le_encoder.transform([y_node_name])[0])  # integer class
    data = Data(x=x, edge_index=edge_index, y=torch.tensor(y_label, dtype=torch.long))
    data_list.append(data)

print("Built", len(data_list), "graph data objects.")

# ---------- 5) Train/val/test split & DataLoaders ----------
# shuffle and split
num_total = len(data_list)
indices = np.arange(num_total)
np.random.seed(42)
np.random.shuffle(indices)
train_end = int(0.7 * num_total)
val_end = int(0.85 * num_total)
train_idx = indices[:train_end]
val_idx = indices[train_end:val_end]
test_idx = indices[val_end:]

train_dataset = [data_list[i] for i in train_idx]
val_dataset = [data_list[i] for i in val_idx]
test_dataset = [data_list[i] for i in test_idx]

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

print("Datasets: train", len(train_dataset), "val", len(val_dataset), "test", len(test_dataset))

# ---------- 6) Define GNN models ----------
class GCNNet(nn.Module):
    def __init__(self, in_channels, hidden_channels, num_classes, num_layers=2):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(GCNConv(in_channels, hidden_channels))
        for _ in range(num_layers-1):
            self.convs.append(GCNConv(hidden_channels, hidden_channels))
        self.lin = nn.Linear(hidden_channels, num_classes)
    def forward(self, x, edge_index, batch):
        for conv in self.convs:
            x = conv(x, edge_index)
            x = F.relu(x)
        x = global_mean_pool(x, batch)   # graph-level embedding
        out = self.lin(x)
        return out

class SAGENet(nn.Module):
    def __init__(self, in_channels, hidden_channels, num_classes, num_layers=2):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(SAGEConv(in_channels, hidden_channels))
        for _ in range(num_layers-1):
            self.convs.append(SAGEConv(hidden_channels, hidden_channels))
        self.lin = nn.Linear(hidden_channels, num_classes)
    def forward(self, x, edge_index, batch):
        for conv in self.convs:
            x = conv(x, edge_index)
            x = F.relu(x)
        x = global_mean_pool(x, batch)
        out = self.lin(x)
        return out

class GATNet(nn.Module):
    def __init__(self, in_channels, hidden_channels, num_classes, num_layers=2, heads=2):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(GATConv(in_channels, hidden_channels, heads=heads))
        for _ in range(num_layers-1):
            self.convs.append(GATConv(hidden_channels*heads, hidden_channels, heads=heads))
        self.lin = nn.Linear(hidden_channels*heads, num_classes)
    def forward(self, x, edge_index, batch):
        for conv in self.convs:
            x = conv(x, edge_index)
            x = F.elu(x)
        x = global_mean_pool(x, batch)
        out = self.lin(x)
        return out

# ---------- 7) Training & eval utilities ----------
def train_epoch(model, loader, optimizer, device):
    model.train()
    total_loss = 0.0
    for batch in loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        out = model(batch.x, batch.edge_index, batch.batch)  # shape (batch_size, num_classes)
        loss = F.cross_entropy(out, batch.y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * batch.num_graphs
    return total_loss / len(loader.dataset)

def evaluate(model, loader, device):
    model.eval()
    all_preds = []
    all_probs = []
    all_labels = []
    with torch.no_grad():
        for batch in loader:
            batch = batch.to(device)
            out = model(batch.x, batch.edge_index, batch.batch)
            probs = F.softmax(out, dim=1).cpu().numpy()
            preds = probs.argmax(axis=1)
            all_probs.append(probs)
            all_preds.append(preds)
            all_labels.append(batch.y.cpu().numpy())
    if len(all_probs) == 0:
        return None
    all_probs = np.vstack(all_probs)
    all_preds = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)
    acc = (all_preds == all_labels).mean()
    return {"accuracy": acc, "preds": all_preds, "probs": all_probs, "labels": all_labels}

# ---------- 8) Train & evaluate each model (GCN, SAGE, GAT) ----------
in_channels = 1         # each node feature is scalar pressure
hidden = 64

models_to_run = {
    "GCN": GCNNet(in_channels, hidden, num_classes).to(DEVICE),
    "GraphSAGE": SAGENet(in_channels, hidden, num_classes).to(DEVICE),
    "GAT": GATNet(in_channels, 64//2, num_classes).to(DEVICE)  # heads inside
}

results_summary = []

for name, model in models_to_run.items():
    print("\n--- Training", name)
    optimizer = optim.Adam(model.parameters(), lr=LR, weight_decay=1e-5)
    best_val_acc = 0.0
    best_state = None
    for epoch in range(1, NUM_EPOCHS+1):
        loss = train_epoch(model, train_loader, optimizer, DEVICE)
        val_metrics = evaluate(model, val_loader, DEVICE)
        val_acc = val_metrics["accuracy"] if val_metrics is not None else 0.0
        print(f"Epoch {epoch:02d} | loss {loss:.4f} | val_acc {val_acc:.4f}")
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_state = {k:v.cpu() for k,v in model.state_dict().items()}
    # restore best
    if best_state is not None:
        model.load_state_dict(best_state)
    # Evaluate on test
    test_metrics = evaluate(model, test_loader, DEVICE)
    test_acc = test_metrics["accuracy"]
    print(f"Model {name} test accuracy: {test_acc:.4f}")
    # Top-k
    probs = test_metrics["probs"]
    labels = test_metrics["labels"]
    def top_k_acc(y_true, y_prob, k):
        topk = np.argsort(y_prob, axis=1)[:, -k:]
        return np.mean([y_true[i] in topk[i] for i in range(len(y_true))])
    top1 = top_k_acc(labels, probs, 1)
    top3 = top_k_acc(labels, probs, 3)
    top5 = top_k_acc(labels, probs, 5)
    print(f"{name} Top-1 {top1:.4f} Top-3 {top3:.4f} Top-5 {top5:.4f}")
    # Save model and results
    torch.save(model.state_dict(), os.path.join(OUT_DIR, f"{name}_gnn.pth"))
    results_summary.append({
        "Model": name,
        "Test_Accuracy": test_acc,
        "Top-1": top1,
        "Top-3": top3,
        "Top-5": top5
    })
    # Save per-model detailed CSV
    pd.DataFrame({
        "label": labels,
        "pred": test_metrics["preds"]
    }).to_csv(os.path.join(OUT_DIR, f"{name}_preds.csv"), index=False)

# ---------- 9) Save summary CSV and plot ----------
df_res = pd.DataFrame(results_summary)
df_res.to_csv(os.path.join(OUT_DIR, "gnn_results_summary.csv"), index=False)

# Bar chart for Top-k
plt.figure(figsize=(8,5))
x = np.arange(len(df_res["Model"]))
w = 0.25
plt.bar(x - w, df_res["Top-1"], width=w, label="Top-1")
plt.bar(x, df_res["Top-3"], width=w, label="Top-3")
plt.bar(x + w, df_res["Top-5"], width=w, label="Top-5")
plt.xticks(x, df_res["Model"])
plt.ylabel("Accuracy")
plt.title("GNN Localization Top-k Results")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "gnn_topk.png"), dpi=300)
plt.close()

print("✅ GNN experiments done. Results saved in", OUT_DIR)
```

```python
# gnn_localization.py
import os
import numpy as np
import pandas as pd
from tqdm import tqdm
import joblib
import matplotlib.pyplot as plt

# PyTorch & PyG imports
import torch
import torch.nn.functional as F
from torch import nn, optim
from torch_geometric.data import Data, DataLoader
from torch_geometric.nn import GCNConv, SAGEConv, GATConv, global_mean_pool

# ---------- Config ----------
DATA_CSV = "net3_dataset_full.csv"
INP_PATH = "Net3.inp"   # used to extract topology via wntr if available
ENCODER_PKL = "leak_node_encoder.pkl"
OUT_DIR = "gnn_results"
BATCH_SIZE = 32
NUM_EPOCHS = 30         # adjust or increase for final experiments
LR = 1e-3
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
TIMESTEPS = 1           # using single snapshot; if using windows, adjust pipeline

os.makedirs(OUT_DIR, exist_ok=True)

# ---------- 1) Build/Load topology (edge_index) ----------
def build_edge_index_from_inp(inp_path):
    try:
        import wntr
    except Exception as e:
        print("wntr not available, cannot parse .inp automatically.", e)
        return None
    wn = wntr.network.WaterNetworkModel(inp_path)
    # build undirected edges from pipes (link endpoint names)
    edges = []
    for pipe_name in wn.pipe_name_list:
        pipe = wn.get_link(pipe_name)
        n1 = pipe.start_node_name
        n2 = pipe.end_node_name
        # convert names to indices as per wn.junction_name_list order
        # we will create mapping later to keep consistent
        edges.append((str(n1), str(n2)))
    return edges, wn.junction_name_list  # return raw edges and node list

# Try to extract topology
edge_pairs = None
node_order = None
try:
    res = build_edge_index_from_inp(INP_PATH)
    if res is not None:
        edge_pairs, node_order = res
        print(f"Loaded topology from {INP_PATH} ({len(edge_pairs)} edges, {len(node_order)} junctions).")
except Exception as e:
    print("Topology extraction failed:", e)

# ---------- 2) Load dataset & encoder ----------
df = pd.read_csv(DATA_CSV)
le_encoder = joblib.load(ENCODER_PKL)
num_classes = len(le_encoder.classes_)
print("Number of classes (nodes):", num_classes)

# Keep only leak samples for localization (we used that previously)
df_loc = df[df["Leak_Label"] == 1].copy()
df_loc["Leak_Node"] = df_loc["Leak_Node"].fillna("Baseline").astype(str)

# Identify node feature columns (pressure columns)
feat_cols = [c for c in df.columns if c not in ["Time","Leak_Node","Leak_Label","Leak_Size"]]
num_nodes = len(feat_cols)
print("Detected node feature columns:", num_nodes)

# Build mapping from node name -> column index:
# We assume your encoder fitted on Leak_Node used the same node names/strings as the column names.
# We will attempt to map encoder.classes_ order to dataset columns: if exact match, use that.
columns_set = set(feat_cols)
mapping_node_to_col = {}
for node_name in le_encoder.classes_:
    if node_name in columns_set:
        mapping_node_to_col[node_name] = feat_cols.index(node_name)

# If we do not have mapping (names differ), fall back to numeric string mapping assuming columns are node IDs
if len(mapping_node_to_col) == 0:
    # Try converting column names and encoder names to strings and match
    for node_name in le_encoder.classes_:
        for idx, col in enumerate(feat_cols):
            if str(col) == str(node_name):
                mapping_node_to_col[node_name] = idx

# If still incomplete, create default mapping using sorted order (last resort)
if len(mapping_node_to_col) != num_nodes:
    print("Warning: mapping between encoder classes and feature columns is incomplete."
          " Using dataset column order as node order. This assumes the encoding matches column order.")
    node_name_order = feat_cols[:]  # string names
else:
    # reorder node list to match columns
    node_name_order = feat_cols[:]  # use columns order

# ---------- 3) Build edge_index (PyG) using node_name_order ----------
# If we have extracted edges with WNTR, map edge node names to indices in node_name_order
if edge_pairs is not None and node_order is not None:
    name_to_idx = {str(n): i for i, n in enumerate(node_name_order)}
    edge_index_list = []
    for a, b in edge_pairs:
        if a in name_to_idx and b in name_to_idx:
            i = name_to_idx[a]
            j = name_to_idx[b]
            edge_index_list.append([i, j])
            edge_index_list.append([j, i])
    edge_index = torch.tensor(edge_index_list, dtype=torch.long).t().contiguous()
    print("Edge index built from INP:", edge_index.shape)
else:
    # fallback: fully connected ring or k-NN? We'll build simple chain edges using column neighbor order
    edge_index_list = []
    for i in range(num_nodes - 1):
        edge_index_list.append([i, i+1])
        edge_index_list.append([i+1, i])
    edge_index = torch.tensor(edge_index_list, dtype=torch.long).t().contiguous()
    print("Fallback edge_index built:", edge_index.shape)

# ---------- 4) Create PyG Data objects (one graph per sample) ----------
data_list = []
labels = []
for idx, row in tqdm(df_loc.iterrows(), total=len(df_loc), desc="Building graphs"):
    # node features: pressures in the order feat_cols
    x = torch.tensor(row[feat_cols].values.astype(np.float32)).unsqueeze(1)  # shape (num_nodes, 1)
    # make graph data
    y_node_name = row["Leak_Node"]
    y_label = int(le_encoder.transform([y_node_name])[0])  # integer class
    data = Data(x=x, edge_index=edge_index, y=torch.tensor(y_label, dtype=torch.long))
    data_list.append(data)

print("Built", len(data_list), "graph data objects.")

# ---------- 5) Train/val/test split & DataLoaders ----------
# shuffle and split
num_total = len(data_list)
indices = np.arange(num_total)
np.random.seed(42)
np.random.shuffle(indices)
train_end = int(0.7 * num_total)
val_end = int(0.85 * num_total)
train_idx = indices[:train_end]
val_idx = indices[train_end:val_end]
test_idx = indices[val_end:]

train_dataset = [data_list[i] for i in train_idx]
val_dataset = [data_list[i] for i in val_idx]
test_dataset = [data_list[i] for i in test_idx]

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

print("Datasets: train", len(train_dataset), "val", len(val_dataset), "test", len(test_dataset))

# ---------- 6) Define GNN models ----------
class GCNNet(nn.Module):
    def __init__(self, in_channels, hidden_channels, num_classes, num_layers=2):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(GCNConv(in_channels, hidden_channels))
        for _ in range(num_layers-1):
            self.convs.append(GCNConv(hidden_channels, hidden_channels))
        self.lin = nn.Linear(hidden_channels, num_classes)
    def forward(self, x, edge_index, batch):
        for conv in self.convs:
            x = conv(x, edge_index)
            x = F.relu(x)
        x = global_mean_pool(x, batch)   # graph-level embedding
        out = self.lin(x)
        return out

class SAGENet(nn.Module):
    def __init__(self, in_channels, hidden_channels, num_classes, num_layers=2):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(SAGEConv(in_channels, hidden_channels))
        for _ in range(num_layers-1):
            self.convs.append(SAGEConv(hidden_channels, hidden_channels))
        self.lin = nn.Linear(hidden_channels, num_classes)
    def forward(self, x, edge_index, batch):
        for conv in self.convs:
            x = conv(x, edge_index)
            x = F.relu(x)
        x = global_mean_pool(x, batch)
        out = self.lin(x)
        return out

class GATNet(nn.Module):
    def __init__(self, in_channels, hidden_channels, num_classes, num_layers=2, heads=2):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(GATConv(in_channels, hidden_channels, heads=heads))
        for _ in range(num_layers-1):
            self.convs.append(GATConv(hidden_channels*heads, hidden_channels, heads=heads))
        self.lin = nn.Linear(hidden_channels*heads, num_classes)
    def forward(self, x, edge_index, batch):
        for conv in self.convs:
            x = conv(x, edge_index)
            x = F.elu(x)
        x = global_mean_pool(x, batch)
        out = self.lin(x)
        return out

# ---------- 7) Training & eval utilities ----------
def train_epoch(model, loader, optimizer, device):
    model.train()
    total_loss = 0.0
    for batch in loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        out = model(batch.x, batch.edge_index, batch.batch)  # shape (batch_size, num_classes)
        loss = F.cross_entropy(out, batch.y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * batch.num_graphs
    return total_loss / len(loader.dataset)

def evaluate(model, loader, device):
    model.eval()
    all_preds = []
    all_probs = []
    all_labels = []
    with torch.no_grad():
        for batch in loader:
            batch = batch.to(device)
            out = model(batch.x, batch.edge_index, batch.batch)
            probs = F.softmax(out, dim=1).cpu().numpy()
            preds = probs.argmax(axis=1)
            all_probs.append(probs)
            all_preds.append(preds)
            all_labels.append(batch.y.cpu().numpy())
    if len(all_probs) == 0:
        return None
    all_probs = np.vstack(all_probs)
    all_preds = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)
    acc = (all_preds == all_labels).mean()
    return {"accuracy": acc, "preds": all_preds, "probs": all_probs, "labels": all_labels}

# ---------- 8) Train & evaluate each model (GCN, SAGE, GAT) ----------
in_channels = 1         # each node feature is scalar pressure
hidden = 64

models_to_run = {
    "GCN": GCNNet(in_channels, hidden, num_classes).to(DEVICE),
    "GraphSAGE": SAGENet(in_channels, hidden, num_classes).to(DEVICE),
    "GAT": GATNet(in_channels, 64//2, num_classes).to(DEVICE)  # heads inside
}

results_summary = []

for name, model in models_to_run.items():
    print("\n--- Training", name)
    optimizer = optim.Adam(model.parameters(), lr=LR, weight_decay=1e-5)
    best_val_acc = 0.0
    best_state = None
    for epoch in range(1, NUM_EPOCHS+1):
        loss = train_epoch(model, train_loader, optimizer, DEVICE)
        val_metrics = evaluate(model, val_loader, DEVICE)
        val_acc = val_metrics["accuracy"] if val_metrics is not None else 0.0
        print(f"Epoch {epoch:02d} | loss {loss:.4f} | val_acc {val_acc:.4f}")
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_state = {k:v.cpu() for k,v in model.state_dict().items()}
    # restore best
    if best_state is not None:
        model.load_state_dict(best_state)
    # Evaluate on test
    test_metrics = evaluate(model, test_loader, DEVICE)
    test_acc = test_metrics["accuracy"]
    print(f"Model {name} test accuracy: {test_acc:.4f}")
    # Top-k
    probs = test_metrics["probs"]
    labels = test_metrics["labels"]
    def top_k_acc(y_true, y_prob, k):
        topk = np.argsort(y_prob, axis=1)[:, -k:]
        return np.mean([y_true[i] in topk[i] for i in range(len(y_true))])
    top1 = top_k_acc(labels, probs, 1)
    top3 = top_k_acc(labels, probs, 3)
    top5 = top_k_acc(labels, probs, 5)
    print(f"{name} Top-1 {top1:.4f} Top-3 {top3:.4f} Top-5 {top5:.4f}")
    # Save model and results
    torch.save(model.state_dict(), os.path.join(OUT_DIR, f"{name}_gnn.pth"))
    results_summary.append({
        "Model": name,
        "Test_Accuracy": test_acc,
        "Top-1": top1,
        "Top-3": top3,
        "Top-5": top5
    })
    # Save per-model detailed CSV
    pd.DataFrame({
        "label": labels,
        "pred": test_metrics["preds"]
    }).to_csv(os.path.join(OUT_DIR, f"{name}_preds.csv"), index=False)

# ---------- 9) Save summary CSV and plot ----------
df_res = pd.DataFrame(results_summary)
df_res.to_csv(os.path.join(OUT_DIR, "gnn_results_summary.csv"), index=False)

# Bar chart for Top-k
plt.figure(figsize=(8,5))
x = np.arange(len(df_res["Model"]))
w = 0.25
plt.bar(x - w, df_res["Top-1"], width=w, label="Top-1")
plt.bar(x, df_res["Top-3"], width=w, label="Top-3")
plt.bar(x + w, df_res["Top-5"], width=w, label="Top-5")
plt.xticks(x, df_res["Model"])
plt.ylabel("Accuracy")
plt.title("GNN Localization Top-k Results")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "gnn_topk.png"), dpi=300)
plt.close()

print("✅ GNN experiments done. Results saved in", OUT_DIR)
```

```python
import os
import numpy as np
import pandas as pd
import joblib
import torch
import torch.nn.functional as F
from torch import nn
from torch_geometric.data import Data, DataLoader
from torch_geometric.nn import GCNConv, SAGEConv, GATConv, global_mean_pool
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
from sklearn.metrics import accuracy_score

# --- Load test data ---
X_test = np.load("X_test.npy")
y_test = np.load("y_test_localization.npy")

# Load encoder
encoder = joblib.load("leak_node_encoder.pkl")
num_classes = len(encoder.classes_)

# --- Noise injection helper ---
def add_noise(X, snr_db):
    if snr_db == np.inf:  # clean
        return X
    signal_power = np.mean(X ** 2)
    snr_linear = 10 ** (snr_db / 10.0)
    noise_power = signal_power / snr_linear
    noise = np.random.normal(0, np.sqrt(noise_power), X.shape)
    return X + noise

# --- Sequence helper for DL models ---
def create_sequences(X, y, timesteps=10):
    Xs, ys = [], []
    for i in range(len(X) - timesteps):
        Xs.append(X[i:(i+timesteps)])
        ys.append(y[i+timesteps])
    return np.array(Xs), np.array(ys)

# --- Load ML/DL/Transformer models ---
models = {
    "RandomForest": joblib.load("rf_localization_model.pkl"),
    "XGBoost": joblib.load("xgb_localization_model.pkl"),
    "LightGBM": joblib.load("lgb_localization_model.pkl"),
    "MLP": load_model("mlp_localization_model.h5"),
    "LSTM": load_model("lstm_localization_model.h5"),
    "CNN": load_model("cnn_localization_model.h5"),
    "Transformer": load_model("transformer_localization_model.h5"),
}

# --- Correct GNN architectures (from Step 23 training) ---
class GCNNet(nn.Module):
    def __init__(self, in_channels, hidden_channels, num_classes, num_layers=2):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(GCNConv(in_channels, hidden_channels))
        for _ in range(num_layers-1):
            self.convs.append(GCNConv(hidden_channels, hidden_channels))
        self.lin = nn.Linear(hidden_channels, num_classes)
    def forward(self, x, edge_index, batch):
        for conv in self.convs:
            x = conv(x, edge_index)
            x = F.relu(x)
        x = global_mean_pool(x, batch)
        return self.lin(x)

class SAGENet(nn.Module):
    def __init__(self, in_channels, hidden_channels, num_classes, num_layers=2):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(SAGEConv(in_channels, hidden_channels))
        for _ in range(num_layers-1):
            self.convs.append(SAGEConv(hidden_channels, hidden_channels))
        self.lin = nn.Linear(hidden_channels, num_classes)
    def forward(self, x, edge_index, batch):
        for conv in self.convs:
            x = conv(x, edge_index)
            x = F.relu(x)
        x = global_mean_pool(x, batch)
        return self.lin(x)

class GATNet(nn.Module):
    def __init__(self, in_channels, hidden_channels, num_classes, num_layers=2, heads=2):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(GATConv(in_channels, hidden_channels, heads=heads))
        for _ in range(num_layers-1):
            self.convs.append(GATConv(hidden_channels*heads, hidden_channels, heads=heads))
        self.lin = nn.Linear(hidden_channels*heads, num_classes)
    def forward(self, x, edge_index, batch):
        for conv in self.convs:
            x = conv(x, edge_index)
            x = F.elu(x)
        x = global_mean_pool(x, batch)
        return self.lin(x)

# --- Load trained GNN weights ---
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
gnn_models = {
    "GCN": GCNNet(1, 64, num_classes).to(DEVICE),
    "GraphSAGE": SAGENet(1, 64, num_classes).to(DEVICE),
    "GAT": GATNet(1, 32, num_classes).to(DEVICE),
}
for name in gnn_models:
    gnn_models[name].load_state_dict(torch.load(f"gnn_results/{name}_gnn.pth", map_location=DEVICE))

# --- Load edge_index from training (must be saved during Step 23) ---
if os.path.exists("gnn_results/edge_index.pt"):
    edge_index = torch.load("gnn_results/edge_index.pt")
else:
    raise FileNotFoundError("Missing gnn_results/edge_index.pt. Please save edge_index during training.")

# --- Noise robustness evaluation ---
snr_levels = [np.inf, 30, 20, 10, 0]
results = []

for snr in snr_levels:
    print(f"\n=== Evaluating at SNR={snr} dB ===")
    X_noisy = add_noise(X_test, snr)

    # --- ML/DL/Transformer ---
    for model_name, model in models.items():
        if model_name in ["LSTM","CNN","Transformer"]:
            X_seq, y_seq = create_sequences(X_noisy, y_test, timesteps=10)
            y_prob = model.predict(X_seq)
            y_pred = np.argmax(y_prob, axis=1)
            y_true = y_seq
        elif hasattr(model, "predict_proba"):
            y_true = y_test
            y_pred = model.predict(X_noisy)
        else:
            y_true = y_test
            y_pred = np.argmax(model.predict(X_noisy), axis=1)

        acc = accuracy_score(y_true, y_pred)
        results.append({"Model": model_name, "SNR": snr, "Accuracy": acc})

    # --- GNNs ---
    data_list = []
    for i in range(len(X_noisy)):
        x = torch.tensor(X_noisy[i].astype(np.float32)).unsqueeze(1)
        y = torch.tensor(y_test[i], dtype=torch.long)
        data = Data(x=x, edge_index=edge_index, y=y)
        data_list.append(data)
    loader = DataLoader(data_list, batch_size=32, shuffle=False)

    for model_name, model in gnn_models.items():
        model.eval()
        preds, labels = [], []
        with torch.no_grad():
            for batch in loader:
                batch = batch.to(DEVICE)
                out = model(batch.x, batch.edge_index, batch.batch)
                pred = out.argmax(dim=1).cpu().numpy()
                preds.extend(pred)
                labels.extend(batch.y.cpu().numpy())
        acc = accuracy_score(labels, preds)
        results.append({"Model": model_name, "SNR": snr, "Accuracy": acc})

# --- Save results ---
df_results = pd.DataFrame(results)
df_results.to_csv("Noise_Analysis_WithGNN.csv", index=False)
df_results.to_excel("Noise_Analysis_WithGNN.xlsx", index=False)

# --- Plot ---
plt.figure(figsize=(10,6))
for model_name in df_results["Model"].unique():
    subset = df_results[df_results["Model"] == model_name]
    plt.plot(subset["SNR"], subset["Accuracy"], marker="o", label=model_name)

plt.xlabel("SNR (dB)")
plt.ylabel("Localization Accuracy")
plt.title("Localization Accuracy vs. Noise Level (Including GNNs)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("Noise_Analysis_WithGNN.png", dpi=300)
plt.close()

print("✅ Noise robustness (with GNNs) complete")
print("✅ Figure: Noise_Analysis_WithGNN.png")
print("✅ Data: Noise_Analysis_WithGNN.csv / Noise_Analysis_WithGNN.xlsx")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib
from sklearn.metrics import accuracy_score
from tensorflow.keras.models import load_model
import random

# === Setup output folder ===
OUT_DIR = "Robustness_Results"
os.makedirs(OUT_DIR, exist_ok=True)

# === Load data ===
X_test = np.load("X_test.npy")
y_test = np.load("y_test_localization.npy")

# === Load models ===
models = {
    "RandomForest": joblib.load("rf_localization_model.pkl"),
    "XGBoost": joblib.load("xgb_localization_model.pkl"),
    "LightGBM": joblib.load("lgb_localization_model.pkl"),
    "MLP": load_model("mlp_localization_model.h5"),
    "LSTM": load_model("lstm_localization_model.h5"),
    "CNN": load_model("cnn_localization_model.h5"),
    "Transformer": load_model("transformer_localization_model.h5"),
}

# === Helper: convert probs → labels ===
def to_labels(y_pred):
    if isinstance(y_pred, np.ndarray) and y_pred.ndim > 1 and y_pred.shape[1] > 1:
        return np.argmax(y_pred, axis=1)
    return y_pred

# === Helper: sequence creation (for DL temporal models) ===
def create_sequences(X, y, timesteps=10):
    Xs, ys = [], []
    for i in range(len(X) - timesteps):
        Xs.append(X[i:(i+timesteps)])
        ys.append(y[i+timesteps])
    return np.array(Xs), np.array(ys)

# === Helper: feature masking for reduced sensors ===
def mask_features(X, selected_sensors, total_features):
    X_masked = np.zeros((X.shape[0], total_features))
    X_masked[:, selected_sensors] = X[:, selected_sensors]
    return X_masked

# ---------------------------------------------------
# 1. Leak Size Robustness (simulate by filtering y_test)
# ---------------------------------------------------
# Example bins (small vs medium vs large leaks)
# Adjust these thresholds based on your dataset labels
leak_bins = {"Small": [0.001], "Medium": [0.005], "Large": [0.01]}
results_size = []

for size_label, sizes in leak_bins.items():
    mask = np.isin(y_test, y_test)  # placeholder (update if you have leak size labels)
    X_sub, y_sub = X_test[mask], y_test[mask]
    
    for model_name, model in models.items():
        if model_name in ["LSTM","CNN","Transformer"]:
            X_seq, y_seq = create_sequences(X_sub, y_sub)
            y_prob = model.predict(X_seq, verbose=0)
            y_pred = to_labels(y_prob)
            acc = accuracy_score(y_seq, y_pred)
        else:
            y_pred = model.predict(X_sub)
            y_pred = to_labels(y_pred)
            acc = accuracy_score(y_sub, y_pred)
        results_size.append({"Model": model_name, "LeakSize": size_label, "Accuracy": acc})

df_size = pd.DataFrame(results_size)
df_size.to_csv(os.path.join(OUT_DIR, "LeakSize_Analysis.csv"), index=False)
df_size.to_excel(os.path.join(OUT_DIR, "LeakSize_Analysis.xlsx"), index=False)

plt.figure(figsize=(10,6))
for model_name in df_size["Model"].unique():
    subset = df_size[df_size["Model"] == model_name]
    plt.plot(subset["LeakSize"], subset["Accuracy"], marker="o", label=model_name)
plt.title("Leak Size Robustness")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "LeakSize_Analysis.png"), dpi=300)
plt.close()

# ---------------------------------------------------
# 2. Sensor Coverage Robustness
# ---------------------------------------------------
coverage_levels = [1.0, 0.8, 0.6, 0.4, 0.2]
results_cov = []

for coverage in coverage_levels:
    num_sensors = int(X_test.shape[1] * coverage)
    selected_sensors = sorted(random.sample(range(X_test.shape[1]), num_sensors))
    X_reduced = mask_features(X_test, selected_sensors, X_test.shape[1])

    for model_name, model in models.items():
        if model_name in ["LSTM","CNN","Transformer"]:
            X_seq, y_seq = create_sequences(X_reduced, y_test)
            y_prob = model.predict(X_seq, verbose=0)
            y_pred = to_labels(y_prob)
            acc = accuracy_score(y_seq, y_pred)
        else:
            y_pred = model.predict(X_reduced)
            y_pred = to_labels(y_pred)
            acc = accuracy_score(y_test, y_pred)

        results_cov.append({"Model": model_name, "Coverage": int(coverage*100), "Accuracy": acc})

df_cov = pd.DataFrame(results_cov)
df_cov.to_csv(os.path.join(OUT_DIR, "SensorReduction_Analysis.csv"), index=False)
df_cov.to_excel(os.path.join(OUT_DIR, "SensorReduction_Analysis.xlsx"), index=False)

plt.figure(figsize=(10,6))
for model_name in df_cov["Model"].unique():
    subset = df_cov[df_cov["Model"] == model_name]
    plt.plot(subset["Coverage"], subset["Accuracy"], marker="o", label=model_name)
plt.title("Sensor Coverage Robustness")
plt.xlabel("Sensor Coverage (%)")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "SensorReduction_Analysis.png"), dpi=300)
plt.close()

# ---------------------------------------------------
# 3. Noise Robustness
# ---------------------------------------------------
def add_noise(X, snr_db):
    if snr_db == np.inf:
        return X
    signal_power = np.mean(X ** 2)
    snr_linear = 10 ** (snr_db / 10.0)
    noise_power = signal_power / snr_linear
    noise = np.random.normal(0, np.sqrt(noise_power), X.shape)
    return X + noise

snr_levels = [np.inf, 30, 20, 10, 0]
results_noise = []

for snr in snr_levels:
    X_noisy = add_noise(X_test, snr)

    for model_name, model in models.items():
        if model_name in ["LSTM","CNN","Transformer"]:
            X_seq, y_seq = create_sequences(X_noisy, y_test)
            y_prob = model.predict(X_seq, verbose=0)
            y_pred = to_labels(y_prob)
            acc = accuracy_score(y_seq, y_pred)
        else:
            y_pred = model.predict(X_noisy)
            y_pred = to_labels(y_pred)
            acc = accuracy_score(y_test, y_pred)

        results_noise.append({"Model": model_name, "SNR": snr, "Accuracy": acc})

df_noise = pd.DataFrame(results_noise)
df_noise.to_csv(os.path.join(OUT_DIR, "Noise_Analysis.csv"), index=False)
df_noise.to_excel(os.path.join(OUT_DIR, "Noise_Analysis.xlsx"), index=False)

plt.figure(figsize=(10,6))
for model_name in df_noise["Model"].unique():
    subset = df_noise[df_noise["Model"] == model_name]
    plt.plot(subset["SNR"], subset["Accuracy"], marker="o", label=model_name)
plt.title("Noise Robustness")
plt.xlabel("SNR (dB)")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "Noise_Analysis.png"), dpi=300)
plt.close()

print("✅ Robustness experiments complete! Files saved in:", OUT_DIR)
```

```python
# Jupyter-ready: Winners by Metric (colored bars, separate legend)
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# --- Load data ---
excel_file = "Results_Tables_Percent.xlsx"   # adjust path if needed
sheet = "Winners"
df = pd.read_excel(excel_file, sheet_name=sheet)

# Expecting columns: ["Metric", "Best Model", "Value"] where Value is like "77.3%"
# Convert Value to numeric percent (float)
def pct_to_float(x):
    if isinstance(x, str) and x.strip().endswith('%'):
        return float(x.strip().strip('%'))
    try:
        return float(x)
    except:
        return np.nan

df["ValuePct"] = df["Value"].apply(pct_to_float)

# --- Build color map for best models ---
models = df["Best Model"].unique()
cmap = plt.get_cmap("tab10")  # good for distinct categories; change if >10 models
colors = {model: cmap(i % cmap.N) for i, model in enumerate(models)}

# Assign bar colors based on the Best Model for each metric
bar_colors = [colors[m] for m in df["Best Model"]]

# --- Plot ---
plt.rcParams.update({"font.size": 12})
fig, ax = plt.subplots(figsize=(11,6))

x = np.arange(len(df))
bars = ax.bar(x, df["ValuePct"], color=bar_colors, edgecolor="k", linewidth=0.6)

# Axes, title, ticks
ax.set_ylabel("Score (%)", fontsize=13)
ax.set_xlabel("Metric", fontsize=13)
ax.set_title("Figure: Winners by Metric (Best model per metric)", fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(df["Metric"], rotation=35, ha="right")

# Remove value labels on top of bars (by user request) -> do nothing here

# --- Legend: show model -> color mapping separately ---
legend_patches = [mpatches.Patch(color=colors[m], label=m) for m in models]
ax.legend(handles=legend_patches, title="Winning Model", bbox_to_anchor=(1.02, 1), loc="upper left")

# Grid and layout
ax.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()

# Save figure
out_fname = "Winners_Summary_Percent_v2.png"
plt.savefig(out_fname, dpi=300, bbox_inches="tight")
plt.show()

print(f"✅ Saved figure: {out_fname}")
```
